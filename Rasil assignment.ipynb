```python
name = input("\nEnter your name: ")
message = input("Enter your favourite quote: ")
code = input("Enter a username or number: ")

print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
print("Title Case:", message.title())
print("Capitalize:", message.capitalize())
print("Swap Case:", message.swapcase())
print("Strip:", name.strip())
print("Left Strip:", name.lstrip())
print("Right Strip:", name.rstrip())
print("Replace:", message.replace("a", "@"))
print("Split:", message.split())
print("Join:", "_".join(message.split()))
print("Find:", message.find("a"))
print("Index:", message.index("a") if "a" in message else "Not found")
print("Count:", message.count("a"))
print("Starts with:", message.startswith("The"))
print("Ends with:", message.endswith("."))
print("Only Alphabets:", name.replace(" ", "").isalpha())
print("Only Digits:", code.isdigit())
print("Alphanumeric:", code.isalnum())
print("Only Spaces:", message.isspace())
print("Is Lowercase:", message.islower())
print("Is Uppercase:", name.isupper())
print("Is Title Case:", name.istitle())
print("Center:", name.center(30, "*"))
print("Left Align:", name.ljust(30, "-"))
print("Right Align:", name.rjust(30, "-"))
print("Zero Fill:", code.zfill(10))
print("Partition:", message.partition(" "))
print("Right Partition:", message.rpartition(" "))
print("Casefold:", name.casefold())
```
